**Task 1: Exploring and Visualizing the Iris Dataset**

**Problem Statement and Goal**





The objective of this task is to explore and visualize the Iris dataset using Python libraries such as Pandas, Matplotlib, and Seaborn. The analysis will help understand the dataset's structure, feature distributions, relationships among variables, and potential outliers.

**Step 1: Import Libraries**

```python
# Data Handling
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris Dataset
from sklearn.datasets import load_iris

# Display plots inline
%matplotlib inline
```

**Step 2: Load Dataset**

**Method 1: Load from Scikit-Learn**

```python
iris = load_iris()

df = pd.DataFrame(iris.data,columns=iris.feature_names)

# Add target column
df['species'] = iris.target

# Convert numeric target to species names
df['species'] = df['species'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})

print("Dataset Loaded Successfully")
```

**Step 3: Inspect Dataset**

**Shape of Dataset**

```python
print("Shape of Dataset:")
print(df.shape)
```

**Column Names**

```python
print("\nColumn Names:")
print(df.columns)
```

**First Five Rows**

```python
print("\nFirst Five Rows:")
print(df.head())
```

**Step 4: Dataset Information**

```python
print("\nDataset Information:")
df.info()
```

**Summary Statistics**

```python
print("\nSummary Statistics:")
print(df.describe())
```

**Step 5: Check Missing Values**

```python
print("Missing Values:")
print(df.isnull().sum())
```

**Data Visualization**


**1. Scatter Plot**

Relationship between Sepal Length and Sepal Width

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='sepal width (cm)',
    hue='species',
    s=100
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
```

**Observation**


1. Setosa is clearly separated from the other species.

2. Versicolor and Virginica have some overlap.




**2. Histograms**


Value and Feature Distributions

```python
features = df.columns[:-1]

fig, axes = plt.subplots(1, 4, figsize=(12, 4))

for i, feature in enumerate(features):
    sns.histplot(
        data=df,
        x=feature,
        kde=True,
        ax=axes[i]
    )
    axes[i].set_title(f"Distribution of {feature}")

plt.tight_layout()
plt.show()
```

 **3.Box Plots**

To Identify Outliers

```python
plt.figure(figsize=(10,6))

sns.boxplot(data=df.iloc[:, :-1])

plt.title("Box Plot of Iris Features")
plt.show()
```
